```python
import pandas as pd
import numpy as np

print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
```

```python
employees = pd.read_csv(data_path / "employees.csv")
jobs = pd.read_csv(data_path / "jobs.csv")
compensation = pd.read_csv(data_path / "compensation.csv")
salary_bands = pd.read_csv(data_path / "salary_bands.csv")
performance = pd.read_csv(data_path / "performance.csv")
promotions = pd.read_csv(data_path / "promotions.csv")

print("All six datasets loaded successfully.")
```

```python
datasets = {
    "Employees": employees,
    "Jobs": jobs,
    "Compensation": compensation,
    "Salary Bands": salary_bands,
    "Performance": performance,
    "Promotions": promotions
}

dataset_summary = pd.DataFrame([
    {
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1]
    }
    for name, df in datasets.items()
])

display(dataset_summary)
```

```python
## Identify Numerical and Categorical Columns

for name, df in datasets.items():

    numerical_cols = df.select_dtypes(
        include=["int64", "float64"]
    ).columns.tolist()

    categorical_cols = df.select_dtypes(
        include=["object", "str", "category"]
    ).columns.tolist()

    print(f"\n{name} Dataset")
    print("Numerical Columns:", numerical_cols)
    print("Categorical/Text Columns:", categorical_cols)
```

## Initial Data Understanding — Key Observations

### 1. Dataset Overview
- The project contains six interconnected datasets covering employees, jobs, compensation, salary bands, performance, and promotions.
- The Employees dataset contains 10,000 records, while the Jobs dataset contains 50 job definitions across 12 departments.
- Compensation and performance data cover three reporting years: 2024–2026.
- Salary bands are defined for different jobs and locations.

### 2. Workforce Characteristics
- The average employee age is 34.06 years.
- Average professional experience is 11.16 years.
- Average employee tenure is 6.41 years.
- Employees are distributed across six Indian cities and six organizational job levels.

### 3. Compensation Characteristics
- Average annual base salary is approximately ₹19.78 lakh.
- Median annual base salary is approximately ₹15.05 lakh.
- Average annual total compensation is approximately ₹22.84 lakh.
- All compensation values are recorded in INR.

### 4. Performance and Promotions
- The average performance rating is 3.20 out of 5.
- The Promotions dataset contains 2,737 promotion records.
- The average recorded promotion-related salary increase is approximately 14.54%.

### 5. Data Structure and Relationships
- Employee_ID connects employees with compensation, performance, and promotions.
- Job_ID connects employees with job definitions.
- Job_ID and Location together identify applicable salary bands.
- Compensation and performance records are organized by employee and reporting year.

### 6. Initial Data Quality Considerations
- Some fields contain missing values, including Education_Level, Years_Experience, Annual_Bonus, and Performance_Rating.
- Date fields require appropriate conversion before time-based analysis.
- Compensation records should be checked for duplicate employee-year combinations.
- Historical compensation and performance records should be reviewed against employee hire dates.
- These observations will be investigated in the Data Quality Audit notebook.

### Summary
The datasets provide a suitable foundation for workforce compensation, pay equity, salary benchmarking, performance, and promotion analysis. Further data validation is required before cleaning, feature engineering, and SQL integration.
